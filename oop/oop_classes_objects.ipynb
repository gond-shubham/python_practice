```python
class book:
    pass
classmate=book()
exercise = book()

class laptop:
    pass
hp=laptop()
lenovo = laptop()

class mobile:
    pass

oneplus= mobile()
iphone= mobile()

print(classmate == exercise)
print(hp == lenovo)
print(oneplus == iphone)
print()
print(classmate is exercise)
print(hp is lenovo)
print(oneplus is iphone)
```
