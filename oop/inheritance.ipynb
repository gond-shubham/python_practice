```python
class Employee:
    def __init__(self, emp_no, amount):
        self.id=emp_no
        self.salary=amount


class Manager(Employee):
    def __init__(self, n, s,d):
        super().__init__(n,s)
        self.department=d


m1=Manager(1221, 100000, "HR")
print(m1.id)

print(m1.__dict__)

#parent object

e1=Employee(1222, 100000)

print(type(e1), type(m1))

print(isinstance(m1, Manager))
print(isinstance(m1, Employee))  #child class object is also a object of parent class.

print(isinstance(e1, Manager)) 
print(isinstance(e1, Employee))

print(issubclass(Manager, Employee))  # Manager is a subclass of a Employee
```

```python
#method overriding.

class Bank:
    name="AXIS"
    CEO="Nithin"

    @classmethod
    def display(cls):
        print(cls)
        print(f'{cls.name}, {cls.CEO}')


class Branch(Bank):
    def __init__(self, Loc, p):
        self.location=Loc
        self.pin=p

    def display(self):
        super().display()    #Here, cls is Branch(<class '__main__.Branch'>).
        print(f'{self.location}, {self.pin}')


branch1=Branch("Adilabad", 504308)

branch1.display()
B=Bank()
print(10*"*")
B.display()

print(Branch.mro())


        
        
    
```

What is Inheritance?
Inheritance is an OOP concept in which one class (child/subclass) acquires the properties and methods of another class (parent/superclass).
Instead of writing the same c

To reuse existing code.
To avoid writing duplicate code.
To make programs easier to maintain.
To extend the functionality of an existing class.
To represent real-world relationships.

Method Overriding
Method overriding means writing a method in the child class with the same name  as a method in the parent class.
When the method is called using the child object, the child's version executes instead of the parent's version.
When we call the method with parent object. The method in  parent class get executed.

super() is used to access the parent class from the child class.
It is commonly used to:
Call the parent class constructor (__init__)
Call parent class methods
Extend the parent's functionality without rewriting it


Method Resolution Order (MRO)
When multiple classes contain the same method, Python follows a specific order to find the method.
This order is called Method Resolution Order (MRO).
