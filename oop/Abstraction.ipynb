```python
import abc

class Animal(abc.ABC):
    
    @abc.abstractmethod
    def sound(self):
        pass


class Cat(Animal):
    
    def sound(self):
        print("meow")


class Dog(Animal):

    def sound(self):
        print("bark")



c1=Cat()
d1=Dog()

c1.sound() # Since Cat and Dog implement (override) the abstract method 'sound()',
d1.sound() # they become concrete classes and objects of these classes can be created.

#a1=Animal() 

print(Animal.__abstractmethods__)  # frozenset({'sound'}) -> because of the abstract method we cann't create object of abstract class.


        

    

```

```python
import abc
import math

class Shape(abc.ABC):
    
    @abc.abstractmethod
    def area (self):
        pass

    @abc.abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radi):
        self.radius=radi

    def area(self):
        return math.pi *self.radius **2

    def perimeter(self):
        return 2*math.pi *self.radius



class Rectangle(Shape):
    def __init__(self,l,b):
        self.len=l
        self.brth=b
        
    def area(self):
        return self.len*self.brth
        
    def perimeter(self):
        return 2*(self.len+self.brth)


c1= Circle(5)

r1=Rectangle(10, 5)

l=[c1,r1]

for obj in l:
    print(obj.area())
    print(obj.perimeter())


```

1. What is Abstraction?
Abstraction is the process of exposing only the essential features (specifications) while hiding the implementation details. The specifications define what an object should do, and the implementation defines how it does it. As end users, we interact with the specifications, while the implementation remains hidden.


2. Why do we use Abstract Classes?
Abstract classes are used to implement abstraction. They define a common interface (specifications) that all subclasses must follow. They are not meant to create objects directly but to act as a blueprint for subclasses, ensuring consistency and supporting polymorphism.


3. What is an Abstract Method?
An abstract method is a method declared in an abstract class without an implementation. Every concrete subclass must implement (override) the abstract method; otherwise, it also becomes an abstract class.

5. Difference between Abstraction and Encapsulation
Abstraction	                                                                       Encapsulation
Hides implementation details.	                                              Hides data from direct access.
Focuses on what an object does.            	                                  Focuses on how data is protected.
Achieved using abstract classes and interfaces.	                              Achieved using classes and access modifiers (public, protected, private).
User sees only the required functionality.	                                  User cannot directly modify sensitive data.
