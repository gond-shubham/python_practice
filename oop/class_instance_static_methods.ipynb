```python
#Instance method
class Student:
    def __init__(self, n, r, m):
        self.name=n
        self.rollno=r
        self.marks=m

    def update_marks(self, mark):
        self.marks+=mark
        
    def display(self):
        return f'{self.name}, {self.rollno}, {self.marks}'

    def check_result(self):
        if self.marks>60:
            return "Pass"

std1=Student("nithin", 101, 70)

res=std1.display()
print(res)


        
```

```python
#shopping cart

class Customer:
    def __init__(self,n):
        self.name=n
        self.cart=[]
    def add_item(self, item):
        self.cart.append(item)
        
    def remove_item(self, item):
        if item in self.cart:
            self.cart.remove(item)
        else:
            print("the item is not available")

    def display(self):
        return self.cart


cust1=Customer("nithin")

cust1.add_item("Laptop")

a=cust1.display()
print(a)

cust1.remove_item("Laptop")

a=cust1.display()
print(a)

        
```
