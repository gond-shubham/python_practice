```python
class Student:
    def __init__(self,n,a): #constructor
        self.name=n
        self.age=a
    def __str__(self):
        n=getattr(self, "name")
        a=getattr(self, "age")
        return f'{n},{a}'

stud1= Student("ramesh", 20)
stud2= Student("suresh", 21)
stud3= Student("naresh", 20)



print(stud1)
print(stud2)
print(stud3)

stud1.age, stud2.age = stud2.age, stud1.age

print(stud1, stud2)


    
```

```python
class Car:
    def __init__(self, b, p):
        self.brand=b
        self.price=p

car1=Car("Audi", 2500000)
car2=Car("Benz", 2600000)

print(car1.brand, car1.price)
print(car2.brand, car2.price)

print(car1, car2) # it will provide object reference without method overiding


```

```python
class Employee:
    def __init__(self,n,i,a):
        self.name=n
        self.id=i
        self.salary=a
    def increase_salary(self, per):
        a=self.salary
        increase=(per*a)/100
        self.salary=a+increase
        print(f'successully increased {increase}')



emp1=Employee("naresh", 101, 40000)
emp2=Employee("suresh", 102, 50000)
emp3=Employee("ramesh", 103, 60000)

emp1.increase_salary(50)
emp2.increase_salary(50)
emp3.increase_salary(50)





```

```python
class BankAccount:
    def __init__(self, n,b):
        self.acc_num=n
        self.balance=b

    def deposit(self,amnt):
        a=self.balance
        self.balance=a+amnt
        
    def withdraw(self, amnt):
        a=self.balance
        if a>amnt:
            self.balance=a-amnt
        else:
            print("insufficient balance")
            



account1= BankAccount(101, 25000)
account2= BankAccount(102, 30000)

print(account1.__dict__)  # obj dict will only contain object attributes.
print(BankAccount.__dict__) # class dict will  contain all the methods and class variables.

account1.deposit(10000)
account2.withdraw(31000) 

print(account1.balance, account2.balance)
```
