```python
class Employee:
    def __init__(self, no, na):
        self.emp_no=no
        self.emp_name=na

    def __str__(self):
        return f'{self.emp_no}, {self.emp_name}'

    def __repr__(self):
        return f'name:{self.emp_name}, id:{self.emp_no}'



e1=Employee(101, "Nithin")
print(e1)
print(repr(e1)) 




#<__main__.Employee object at 0x00000295C8C3AF90> will get object reference without creating __str__ or __repr__
```

```python
class Rectangle:
    def __init__(self,l,b):
        self.len=l
        self.bth=b

    def __eq__(self, other):
        if self.len == other.len and self.bth == other.bth:
            return True
        return False

    def __lt__(self, other):
        if self.len < other.len and self.bth < other.bth:
            return True
        return False
        
    def __gt__(self, other):
        if self.len > other.len and self.bth > other.bth:
            return True
        return False


r1=Rectangle(15, 10)
r2=Rectangle(16, 11)
r3=Rectangle(15, 10)

print(r1==r2)
print(r1==r3)


print(r1<r2)
print(r1>r2)



            
```

1. What are Magic (Dunder) Methods?
-> Magic methods (also called dunder methods, because they have double underscores) are special methods that Python automatically calls when a     specific operation is performed on an object.


2. Why are they called automatically?
Because the Python interpreter has predefined rules. When Python encounters certain syntax, it automatically translates it into a corresponding magic method call.



3. What is Operator Overloading?
Operator overloading means changing or defining the behavior of an operator (+, -, *, ==, etc.) for user-defined objects.



Magic (Dunder) Methods

Special methods (having double underscores) that Python automatically invokes when certain operations are performed on an object.

Why are they called automatically?

The Python interpreter has built-in rules that map language operations (like object creation, printing, arithmetic, comparisons, indexing, etc.) to their corresponding magic methods.

Operator Overloading

Operator overloading is the process of redefining the behavior of existing operators (such as +, -, *, ==) for user-defined classes by implementing the corresponding magic methods like __add__(), __sub__(), and __eq__().

```python
#singleton class
class College:
    def __init__(self,n):
        self.name=n


c1=College("A")
c2=College("B")

#print(id(c1), id(c2))


import sys

class University:
    c=0
    def __init__(self,n):
        if not hasattr(self, "name"):
            self.name=n
        

    def __new__(cls, *args, **kwargs):
        if cls.c==0:
            cls.c=super().__new__(cls)
            return cls.c
        return cls.c


u1=University("ABC")
print(u1.name)
u2=University("AB")
print(u2.name)

print(sys.getrefcount(u1))

'''print(u1, u2, sep="\n")
print(id(u1), id(u2))'''





        
        
        
```



What is a Singleton?
A Singleton is an object for which only one instance exists throughout the program. Every reference points to the same object in memory.

A Singleton Design Pattern is a class designed by the programmer so that only one object of that class can ever be created.
Unlike Python's built-in singleton objects, we implement this behavior ourselves.

1. Python Language-Provided Singleton Objects
These are built into Python. Only one instance of each exists.
None
True
False
NotImplemented

These are created by Python itself. We never create new objects of these types.

Uses
None → Represents "no value" or "missing value".
True / False → Used as Boolean flags and conditions.
NotImplemented → Used in operator overloading to tell Python that an operation is not implemented for the given operands.
