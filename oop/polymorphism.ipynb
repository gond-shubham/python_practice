```python
class Animal:

    @staticmethod
    def sound():
        print("makes sound")



class Dog(Animal):
    @staticmethod
    def sound():
        print("Bark: Booww!")


class Cat(Animal):
    @staticmethod
    def sound():
        print("Meow..!, Meow....!")


class Cow(Animal):
    @staticmethod
    def sound():
        print("Ambaa..!")



d1=Dog()
Dog.sound()
d1.sound()

c1=Cat()
Cat.sound()
c1.sound()

c2=Cow()
Cow.sound()
c2.sound()

a=Animal()
Animal.sound()
a.sound()
        
```

```python
import math
class Shape:
    def area(self):
        print("calculate the area of the shape")


class Rectangle(Shape):
    def __init__(self,l,b):
        self.len=l
        self.bred=b

    def area(self):
        print(self.len*self.bred)


class Circle(Shape):
    def __init__(self,r):
        self.radi=r

    def area(self):
        print(math.pi*self.radi**2)
    
        

r1=Rectangle(10,5)
c1=Circle(10)

List_obj=[r1, c1]

for i in List_obj:
    i.area()
```

```python
#Duck typing

class Bird:
    def fly(self):
        print("the bird is flying")


class Aeroplane:
    def fly(self):
        print("the aeroplane is flying")

def fun_fly(obj):
    obj.fly()


b1=Bird()
fun_fly(b1)

a1=Aeroplane()
fun_fly(a1)


class Drone:
    def fly(self):
        print("the drone is flying")

d1=Drone()
fun_fly(d1)



```

what is ducktyping.  --  "If it walks like a duck and quacks like a duck, treat it as a duck."
- Duck typing is a feature of Python where the type of an object is less important than the methods or behavior it provides.

Advantages of it.
- Very flexible
- Less code
- Easy to extend
- Promotes reusable functions
- No unnecessary inheritanc


we can avoid the issues or error from ducktyping with the help of.
- instance , error handling or abstraction classes. 


```python
class String:
    def __init__(self,st):
        self.s=st
    def __len__(self):
        count=0
        for i in self.s:
            count+=1
        return count


s=String("abc")
print(len(s))
            
        
    
```

```python
class List:
    def __init__(self, *args):
        self.l=[*args]

    def __len__(self):
        count=0
        for i in self.l:
            count+=1
        return count


l=List(1,2,3,4,5,6,7,8,9,0)

print(len(l))
```

```python
class Tuple:
    def __init__(self, *t):
        self.tp=t
        
    def __len__(self):
        count=0
        for i in self.tp:
            count+=1
        return count

t=Tuple(4, 5, 6, "a", "b", 1)

print(len(t))
        
        
```

```python
class Dict:
    def __init__(self, **kwargs):
        self.d=kwargs

    def __len__(self):
        c=0
        for i in self.d:
            c+=1
        return c
        
d={'a':1, 'b':2, 'c':3}

d1=Dict(**d)

print(len(d1))
```

```python

class Int:
    def __init__(self, val):
        self.value=val
        
    def __add__(self, others):
        return Int(self.value +others.value)

    def __radd__(self, others):
        return Int(self.value + others)
        

    def __str__(self):
        return f'{self.value}'


    def __repr__(self):
        return f'User defined class {Int} have a attribute with  value {self.value}'
            

a=Int(5)
b=Int(6)
c=Int(7)
print(a+b)
print(10+a+b+c)


        
```

What is Polymorphism?
Polymorphism means "one interface, many forms."
It allows the same method, function, or operator to perform different actions depending on the object or data type.

Why is Polymorphism useful?
Makes code reusable.
Makes programs flexible and easy to extend.
Reduces duplicate code.
Allows writing generic code that works with different object types.

Method Overriding
Method Overriding is when a child class redefines a method that already exists in the parent class.
The child class provides its own implementation of the method.
