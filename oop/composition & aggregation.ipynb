Composition

Composition is a strong "has-a" relationship in which one object creates and owns another object. The contained object's lifetime depends on the container object.

```python
class Employee:
    def __init__(self, no, sal, name):
        self.emp_no= no
        self.salary =sal
        self.name=name


class Department:
    def __init__(self, dn, di,e1, e2):
        self.dep_name=dn
        self.dep_id=di
        self.emp_1=e1
        self.emp_2=e2



emp1=Employee(1, 50000, "naresh")
emp2=Employee(2,  60000, "suresh")


dep_HR=Department("HR", 101,emp1, emp2)


print(emp1)  # <__main__.Employee object at 0x000001452880B8C0>

del emp1

print(dep_HR.emp_1)  #<__main__.Employee object at 0x000001452880B8C0>

```

Aggregation

Aggregation is a weak "has-a" relationship in which one object uses another existing object. Both objects can exist independently.
