```python
# how the __setattr__ works

class student:
    def __delattr__():
        print("don't delete")

    def __setattr__(self, attribute, value):
        print("don't set")

    def __getattr__(self, attribute):
        return self.attribute




s1=student()

setattr(s1, "name", "ramesh")  
s1.name="ramesh"




        
```

```python
class Calculator:
    @staticmethod
    def add(a,b):
        return a+b

c1=Calculator()
value1=Calculator.add(5,5)
value2= c1.add(9,9)      #static method don't take object reference
print(value1, value2)
```

```python
class BankAccount:
    def display_balance(self):  #instance method
        return self.balance


accon1= BankAccount()
accon2= BankAccount()

accon1.balance=25000
accon2.balance=50000

b1=accon1.display_balance()
b2=accon2.display_balance()

print(b1, b2)
```

```python
class Rectangle:
    def perimeter(self):
        return 2*(self.len+self.brea)
        
    def area(self):
        return self.len*self.brea

rect1=Rectangle()
rect2=Rectangle()

rect1.len=5
rect1.brea=10

rect2.len=9
rect2.brea=18

area=rect1.area()
peri =rect1.perimeter()

print(area, peri)




        
        
```

```python
class Circle:
    def circumference(self):
        return 2*22/7*self.radius



c1=Circle()

c1.radius=10

circumf = c1.circumference()
print(round(circumf,4))
```
