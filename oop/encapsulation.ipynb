```python
class Employee:
    def __init__(self,n,Id, amount):
        self.name=n
        self._sap_id=Id                     #the leading underscore is a signal that it is intended for internal use.
        self.__salary=amount


emp1=Employee("naresh", 200212, 50000)

print(emp1.name)
print(emp1._sap_id)           #we can access outside the class but we shouldn't do.

#print(emp1.__salary)

print(emp1.__dict__)  # {'name': 'naresh', '_sap_id': 200212, '_Employee__salary': 50000}

print(emp1._Employee__salary)  #name mangling python hides the private attributes by changing the names

print(hasattr(emp1, "__salary")) #False

print(hasattr(emp1, "_Employee__salary"))  #True

setattr(emp1, "__salary", 100000)  
print(emp1._Employee__salary)

setattr(emp1, "_Employee__salary", 100000)
print(emp1._Employee__salary)
```

```python
class BankAccount:
    def __init__(self, amount):
        self.balance=amount
        
    @property
    def balance(self):
        return self.__balance
    
    @balance.setter
    def balance(self, amount):
        if 0<=amount:
            self.__balance=amount
        else:
            raise ValueError("balance should be positive")
            
    def withdraw(self, amount):
        if self.balance>amount:
            self.balance-=amount
        else:
            print("Insufficient balance")

    def deposit(self, amount):
        if amount>0:
            self.balance+=amount



            

print(BankAccount.__dict__["balance"])

a1=BankAccount(100)

a1.deposit(100)
print(a1.balance)

a1.withdraw(300)

#a1.balance=-10

print(dir(BankAccount))
print





```

Encapsulation
What is Encapsulation?
Definition:Encapsulation is the process of combining data (variables) and methods (functions) into a single unit (class) and controlling access to the data.

Keep data and related methods together.
Protect important data from direct modification.
